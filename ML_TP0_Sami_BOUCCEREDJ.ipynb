```python
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
```

```python
# 1. Appliquer PCA
pca = PCA(n_components=2)  # Réduction à 2 dimensions
IrisPCA = pca.fit(data).transform(data)

# 2. Appliquer LDA
lda = LDA(n_components=2)  # Réduction à 2 dimensions (on peut aussi faire n_components=1 ou plus)
IrisLDA = lda.fit(data, label).transform(data)

# Convertir les résultats en DataFrame pour mieux visualiser
IrisPCA_df = pd.DataFrame(IrisPCA, columns=['PCA Component 1', 'PCA Component 2'])
IrisLDA_df = pd.DataFrame(IrisLDA, columns=['LDA Component 1', 'LDA Component 2'])

# Afficher les premières lignes des résultats
print("PCA Results:")
print(IrisPCA_df.head())

print("\nLDA Results:")
print(IrisLDA_df.head())
```

```python
# Plot des résultats PCA
plt.figure(figsize=(8, 6))
plt.scatter(IrisPCA[:, 0], IrisPCA[:, 1], c=label, cmap='viridis', edgecolor='k', s=50)
plt.title("Projection des données Iris avec PCA")
plt.xlabel('Composante Principale 1')
plt.ylabel('Composante Principale 2')
plt.xlim(IrisPCA[:, 0].min() - 1, IrisPCA[:, 0].max() + 1)
plt.ylim(IrisPCA[:, 1].min() - 1, IrisPCA[:, 1].max() + 1)
plt.colorbar(label='Classe')  # Légende pour la couleur en fonction des classes
plt.show()

# Plot des résultats LDA
plt.figure(figsize=(8, 6))
plt.scatter(IrisLDA[:, 0], IrisLDA[:, 1], c=label, cmap='viridis', edgecolor='k', s=50)
plt.title("Projection des données Iris avec LDA")
plt.xlabel('Composante LDA 1')
plt.ylabel('Composante LDA 2')
plt.xlim(IrisLDA[:, 0].min() - 1, IrisLDA[:, 0].max() + 1)
plt.ylim(IrisLDA[:, 1].min() - 1, IrisLDA[:, 1].max() + 1)
plt.colorbar(label='Classe')  # Légende pour la couleur en fonction des classes
plt.show()
```

Pour LDA, les classes sont un peu plus séparées que pour PCA. Cela s'explique par le fait que LDA est une méthode supervisée qui essaie de maximiser la séparation entre les classes, alors que PCA cherche plutôt les variables responsables de la variance des données sans se soucier des classes.

```python
from sklearn.manifold import TSNE

# Appliquer t-SNE pour réduire la dimension à 2
tsne = TSNE(n_components=2, random_state=42, perplexity=30)
IrisTSNE = tsne.fit_transform(data)
"""
La perplexité est un paramètre crucial dans le fonctionnement de t-SNE. 
Elle détermine l'équilibre entre la préservation des relations locales et 
globales dans les données :
- Si la perplexité est faible (par exemple 5), t-SNE se concentrera sur la 
préservation des relations très locales (proximité immédiate entre les 
points voisins).
- Si la perplexité est élevée (par exemple 50), t-SNE considérera des 
voisins plus distants.
"""

# Visualisation des résultats de t-SNE
plt.figure(figsize=(8, 6))
plt.scatter(IrisTSNE[:, 0], IrisTSNE[:, 1], c=label, cmap='viridis', edgecolor='k', s=50)
plt.title("Projection des données Iris avec t-SNE")
plt.xlabel("t-SNE Composante 1")
plt.ylabel("t-SNE Composante 2")
plt.colorbar(label="Classe")  # Légende pour la couleur en fonction des classes
plt.show()
```

Observations :
- PCA : Conserve la variance globale des données, mais ne cherche pas à séparer les classes.
- LDA : Supervisé, maximise la séparation entre les classes.
- t-SNE : Gère très bien la séparation des groupes en 2D, notamment pour des structures complexes, car il essaie de préserver les relations de proximité locales entre les points, c'est-à-dire qu'il cherche à ce que les points similaires dans l'espace d'origine restent proches dans l'espace réduit

### C. Clustering 

```python
from sklearn.datasets import make_blobs

# Générer 4 clusters aléatoires pour 300 échantillons dans 2 dimensions
X, y = make_blobs(n_samples=300, centers=4, n_features=2, random_state=42)

# Visualiser les données en 2D en utilisant la couleur des classes
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', edgecolor='k', s=50)
plt.title("Clustering avec make_blobs (4 clusters)")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.colorbar(label="Classe") 
plt.show()
```

```python
X.shape
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=4, random_state=42)
y_kmeans = kmeans.fit_predict(X)

# 3. Visualiser les résultats du clustering KMeans en 2D
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, cmap='viridis', edgecolor='k', s=50)
plt.title("Clustering des données Iris avec KMeans (4 clusters)")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.colorbar(label="Cluster")  # Légende pour la couleur en fonction des clusters
plt.show()
```

Les clusters sont séparés de la même façon avec kmeans (à part un point dans le cluster violet qui appartenait au jaune). Cela montre que kmeans a bien fait le clustering.

```python
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

hierarchical = AgglomerativeClustering(n_clusters=4)
y_hierarchical = hierarchical.fit_predict(X)

# Visualiser les résultats du clustering hiérarchique en 2D
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y_hierarchical, cmap='viridis', edgecolor='k', s=50)
plt.title("Clustering hiérarchique (4 clusters)")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.colorbar(label="Cluster")  # Légende pour la couleur en fonction des clusters
plt.show()

# Visualiser le dendrogramme
linked = linkage(X, 'ward')  # Calculer la matrice de distances pour le dendrogramme

plt.figure(figsize=(10, 7))
dendrogram(linked)
plt.title("Dendrogramme de clustering hiérarchique")
plt.xlabel("Index des points")
plt.ylabel("Distance")
plt.show()
```

```python
from sklearn.metrics import silhouette_score

X, y = make_blobs(n_samples=300, centers=4, n_features=2, random_state=42)

silhouette_scores = []
cluster_range = range(2, 11)  # Tester le nombre de clusters de 2 à 10

for n_clusters in cluster_range:
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    kmeans.fit(X)
    
    # Calculer l'index de silhouette
    score = silhouette_score(X, kmeans.labels_)
    silhouette_scores.append(score)
    
    print(f"Silhouette Score pour {n_clusters} clusters: {score:.4f}")

plt.figure(figsize=(8, 6))
plt.plot(cluster_range, silhouette_scores, marker='o', color='b')
plt.title("Silhouette Scores pour différents nombres de clusters")
plt.xlabel("Nombre de clusters")
plt.ylabel("Silhouette Score")
plt.show()

optimal_clusters = cluster_range[np.argmax(silhouette_scores)]
print(f"Le nombre optimal de clusters est : {optimal_clusters}")
```

### E. Clustering - part A

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from scipy.spatial.distance import squareform

# Exemple de matrice de dissimilarité complète (symétrique)
dissimilarity_matrix = [
    [0.0, 0.3, 0.4, 0.7],
    [0.3, 0.0, 0.5, 0.8],
    [0.4, 0.5, 0.0, 0.45],
    [0.7, 0.8, 0.45, 0.0]
]

# Convertir la matrice de dissimilarité en format condensé (un vecteur)
dissimilarity_vector = squareform(dissimilarity_matrix, checks=False)

# Calculer le lien hiérarchique avec la méthode 'complete'
linked = linkage(dissimilarity_vector, method='complete')

# Visualiser le dendrogramme
plt.figure(figsize=(10, 7))
dendrogram(linked)

# Ajouter des étiquettes et des titres
plt.title("Dendrogramme de clustering hiérarchique (Complete Linkage)")
plt.xlabel("Index des points")
plt.ylabel("Hauteur (distance)")

# Afficher le dendrogramme
plt.show()
```

```python
# Exemple de matrice de dissimilarité complète (symétrique)
dissimilarity_matrix = [
    [0.0, 0.3, 0.4, 0.7],
    [0.3, 0.0, 0.5, 0.8],
    [0.4, 0.5, 0.0, 0.45],
    [0.7, 0.8, 0.45, 0.0]
]

# Convertir la matrice de dissimilarité en format condensé (un vecteur)
dissimilarity_vector = squareform(dissimilarity_matrix, checks=False)

# Calculer le lien hiérarchique avec la méthode 'complete'
linked = linkage(dissimilarity_vector, method='single')

# Visualiser le dendrogramme
plt.figure(figsize=(10, 7))
dendrogram(linked)

# Ajouter des étiquettes et des titres
plt.title("Dendrogramme de clustering hiérarchique (Single Linkage)")
plt.xlabel("Index des points")
plt.ylabel("Hauteur (distance)")

# Afficher le dendrogramme
plt.show()
```
